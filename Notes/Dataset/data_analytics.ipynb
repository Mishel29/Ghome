```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_excel("dataset.xlsx")
df.shape
```

```python
temp_df = df[['Completion Year', 'Price', 'Current Year']]
temp_df
```

```python
temp_df = df.copy()

rng = np.random.default_rng(42)


def generate_price_history(row):
    completion_year = int(row['Completion Year'])
    current_year = int(row['Current Year'])
    final_price = float(row['Price'])

    if completion_year >= current_year:
        return pd.Series({
            'Years': None,
            'Historical Prices': None
        })

    years = list(range(completion_year, current_year + 1))

    prices = [None] * len(years)
    prices[-1] = final_price

    for i in range(len(prices) - 2, -1, -1):
        change = rng.uniform(-0.10, 0.08)

        prices[i] = prices[i + 1] / (1 + change)

    prices = [round(price) for price in prices]

    prices[-1] = round(final_price)

    return pd.Series({
        'Years': years,
        'Historical Prices': prices
    })


temp_df[['Years', 'Historical Prices']] = temp_df.apply(
    generate_price_history,
    axis=1
)

temp_df.head()
```

```python
temp_df.to_excel("data_with_history.xlsx")
```
